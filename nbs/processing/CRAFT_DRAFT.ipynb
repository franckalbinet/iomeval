# IOMEval Project Context

## Overview
IOMEval is a project of IOM's Central Evaluation Division (CED) for processing and tagging
IOM evaluation reports against international migration frameworks. It comprises two main components:
- **IOMEval** (`iomeval/`) — a Python library (nbdev) for the processing pipeline
- **TagApp** (`tagapp/`) — a FastHTML web app (deployed to pla.sh) for CED staff to review/edit tags

## Pipeline: Download → OCR → Curate → Map → Deploy

1. **Data ingestion** (`readers.py`): IOM eval repo CSV → normalized `evaluations.json`, enriched with UNEG API report URLs
2. **Download** (`downloaders.py`): Fetch PDF reports → `data/pdf/{eval_id}/`
3. **OCR** (via `mistocr`): PDF → markdown with heading hierarchy fixing → `data/md/{eval_id}/`
4. **Curation** (`curator.py`): FastHTML app to (i) fix markdown heading hierarchy, (ii) select sections for tagging
5. **Mapping** (`mapper.py`): LLM-based tagging (via litellm) of selected sections against:
   - GCM objectives (Global Compact for Migration)
   - SRF outputs (Strategic Results Framework)
   - SRF enablers

The JSON file is saved after each pipeline step (OCR, curation, each mapping type).

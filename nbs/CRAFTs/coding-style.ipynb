# Coding Style

How to write and architect python code the fastai/fasctore way.

 
## Philosophy
 
fastai style is inspired by APL/J/K traditions: **brevity facilitates reasoning**.
Keep all key ideas for one semantic concept visible on a single screen.
No scrolling. No hunting across files.
 
Style is deliberately non-PEP8. Do not auto-format or lint existing code.
 

## Core Principles

 ### 1. Short, Dense Functions
- One line of code = one complete idea
- Inline `if`/`else` on a single line when body is short
- Consecutive short functions need no blank lines between them

```python
def det_lighting(b, c): return lambda x: lighting(x, b, c)
def det_rotate(deg):    return lambda x: rotate_cv(x, deg)
def det_zoom(zoom):     return lambda x: zoom_cv(x, zoom)
```

### 2. Top-Down / Expository Development
- Start with a working example, refine incrementally
- Code tells *how*; only add comments to say *why*
- Include links to papers/docs inline when implementing them

 
### 3. Literate Programming with `@patch`
Organize code by *concept*, not by *class definition*.
Use `fastcore.basics.patch` to add methods to a class from anywhere:
 
```python
from fastcore.basics import patch
 
@patch
def predict(self: Learner, row):
    ...
```
 
This keeps related behaviors together in notebooks/modules.

### 4. Liberal Use of fastcore Utilities
 
| Utility | Use case |
|---|---|
| `L` | Preferred list type — chainable, functional |
| `store_attr()` | Auto-assigns `__init__` params to `self` |
| `delegates(to)` | Forward kwargs cleanly between functions/classes |
| `Path` (pathlib) | File paths — no spaces around `/` operator |
| `AttrDict` | Dict with attribute access |
| `ifnone(val, default)` | Concise null coalescing |
| `listify(o)` | Ensure something is a list |
| `setify(o)` | Ensure something is a set |
| `merge(*ds)` | Merge dicts |
| `parallel(f, items)` | Easy parallelism with progress |
| `@call_parse` | Turn a function into a CLI script |

 ## Naming Conventions
 
- Classes: `CamelCase` — everything else: `under_score`
- **Huffman coding**: common things get short names
  - `sz` not `size`, `img` not `image`, `tfm` not `transform`
  - `lm` = language model, `kl_divergence` not `kullback_leibler_divergence`
- In comprehensions: `o` = object, `i` = index, `k`/`v` = key/value
- Tensor inputs to layers/transforms: `x`
- Follow paper nomenclature when implementing papers (e.g. `conv1`, not `first_conv_layer`)
- Assume domain knowledge in the reader

 
## Layout Rules
 
- Max line width: ~160 chars (fits a 1600px screen at 14pt)
- Destructuring assignment for multiple attrs — no spaces after commas:
  ```python
  self.sz,self.denorm,self.norm = sz,denorm,normalizer
  ```
- Align conceptually similar lines so differences are scannable:
  ```python
  if self.store.stretch_dir==0: x = stretch_cv(x, self.store.stretch, 0)
  else:                         x = stretch_cv(x, 0, self.store.stretch)
  ```
- Multi-module imports on one line:
  ```python
  import PIL, os, numpy as np, math, collections
  ```
- Use `import *` when using most of a module; control exports via `__all__`
- Avoid trailing whitespace; avoid unnecessary vertical space
 

 ## Algorithms & Performance
 
- Use numpy/PyTorch broadcasting and advanced indexing — no loops
- Must scale to arbitrarily large datasets (use generators/lazy structures)
- Annotate with equation numbers from source paper in comments
- Profile hotspots before submitting; match or beat reference implementations

 
## Development Workflow
 
- Keep PRs small and focused
- Discuss controversial changes on forums.fast.ai first
- Test on multiple datasets before PRing a new technique
- Don't rely on unreleased/beta library versions

 
## Documentation
 
- One-line docstrings with backtick references to key params
- Docs live in notebooks (`docs_src/`), not just docstrings
- Model: Python `re` module documentation style
- Always use fastcore "docments" parameters decription and docstring as below:
```
def add(
    a:int, # the 1st number to add
    b=0,   # the 2nd number to add
)->int:    # the result of adding `a` to `b`
    "The sum of two numbers."
    return a+b
```

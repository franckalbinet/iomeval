Yes! You can use `PyMuPDF` (package name `pymupdf` or `fitz`):

```python
import fitz  # PyMuPDF

def remove_images(input_path, output_path):
    doc = fitz.open(input_path)
    for page in doc:
        for img in page.get_images():
            page.delete_image(img[0])
    doc.save(output_path, garbage=4, deflate=True)
    doc.close()
```

````
def replace_in_pages(pg_dir, old_str, new_str=''):
    "Replace `old_str` with `new_str` in all markdown files in `pg_dir`, in place."
    pg_dir = Path(pg_dir)
    for f in sorted(pg_dir.glob('*.md')):
        txt = f.read_text()
        f.write_text(txt.replace(old_str, new_str))
````

`````
prompt_fix_hdgs_v2 = """You are cleaning markdown headings extracted from OCR'd evaluation reports. These reports follow a general structure (with variation):

TYPICAL STRUCTURE:
- Title (H1 — the ONLY H1)
- Executive Summary (H2)
- Table of Contents (H2) — may appear before or after Executive Summary
- List of Figures / List of Abbreviations (H2)
- Introduction / Background (H2)
- Findings (H2)
- Conclusions (H2)
- Recommendations (H2)
- Annexes (H2)

CRITICAL: DISTINGUISHING TOC ITEMS FROM REAL HEADINGS
Table of Contents entries list the document's sections but are NOT actual section headings. Clues that something is a TOC item (not a real heading):
- It appears between "Table of Contents" and the next major section (e.g., "Introduction")
- Multiple items appear in sequence on the same or consecutive pages with page number references
- The TOC heading text will be REPEATED later as an actual heading
→ TOC items should be converted to regular text: output the text WITHOUT any # symbols

MISCLASSIFIED HEADINGS (output as "" to remove, or without # to convert to text):
- Table titles (e.g., "Table 1: Budget Overview")
- Figure captions (e.g., "Figure 3.1: Migration flows")
- Bold text used for emphasis, not structure
- Items in bulleted/numbered lists misidentified as headings
- Page headers/footers repeated across pages

HEADING HIERARCHY RULES:
1. **Single H1**: Only the document title should be #. Everything else is ## or deeper.

2. **Numbered sections imply hierarchy**:
   - "1. Introduction" (##) → "1.1 Background" (###) → "1.1.1 Context" (####)
   - Parent sections MUST be shallower than children
   - Works with any style: 1/1.1, A/A.1, I/I.A, etc.
   
   *Example error:*
   ```
   ## 3. Section title
   ## 3.1 Subsection title  ← WRONG: should be ### because 3.1 is a child of 3
   ```
   
   The number of segments determines depth:
   - "3" (1 segment) → parent level
   - "3.1" (2 segments) → one level deeper than parent
   - "3.1.1" (3 segments) → two levels deeper than parent

3. **No level jumps**: Going deeper can only increase by one # at a time
   - Wrong: ## → ##### 
   - Right: ## → ###

4. **Unnumbered headings in numbered documents**: If the document uses consistent numbering, an unnumbered "heading" within that structure is likely misclassified bold text → remove the # symbols

5. **Mixed or no numbering**: Use semantic judgment based on content and typical report structure above

ANALYSIS APPROACH:
1. First, identify the title (should be the only H1)
2. Locate the TOC section — everything between "Table of Contents"/"Contents" heading and the next major section is likely TOC items, not headings
3. For each heading, determine: Is this a real structural heading, a TOC item, or misclassified text?
4. For real headings, verify the # depth matches the implied hierarchy from numbering or semantic role
"""
`````

## Monitor batch runs

`````
import json

def load_all_batch_runs(base_path):
    """Load all batch run JSON files from the batch_runs folder."""
    return {p.stem: json.loads(p.read_text()) for p in (base_path / 'batch_runs').ls()}

def get_all_failed(base_path):
    """Return dict of {batch_name: [failed_ids]} for all runs with failures."""
    return {name: br['failed'] for name, br in load_all_batch_runs(base_path).items() if br.get('failed')}

def find_status(base_path, id):
    """Return list of (batch_name, status, error) tuples for an ID across all batch runs."""
    results = []
    for name, br in load_all_batch_runs(base_path).items():
        for status in ['completed', 'awaiting_curation', 'skipped']:
            if id in br.get(status, []):
                results.append((name, status, None))
        for failed in br.get('failed', []):
            if failed.get('id') == id:
                results.append((name, 'failed', failed.get('error')))
    return results
`````

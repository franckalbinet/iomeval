Looking at the `json_tmp` template, here's the structure:

**CodeSets** — each codeset has an `AttributesList` of dimensions. Each dimension is `"Not selectable (no checkbox)"` (a category header) with nested child `Attributes` that are `"Selectable (show checkbox)"` (the actual tags). Key fields per attribute: `AttributeId` (unique), `AttributeSetId` (unique), `AttributeName`.

**References** — each entry links to codes via `Codes: [{AttributeId: ...}]`, plus metadata fields (`ItemId`, `Title`, `Abstract`, `URL`, etc.). `ItemId` is numeric.

So the plan for your data would be:

- **Dimension 1: Country** — leaf values = unique countries across all 20 mapped results
- **Dimension 2: GCM Objective** — leaf values = only GCM objectives appearing with `relevance_score > 0.83`
- **References** — one per mapped result, coded with its countries + qualifying GCM objectives, using the reasoning as abstract and `report_url` as URL
- Generate numeric IDs for everything (to be replaced later)

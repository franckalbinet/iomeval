def build_eppi_expanded(data, y_meta='Regions Covered', y_label='Region',
                        theme='gcms', thresholds=(0.66, 0.84),
                        seg_labels=('Relevant', 'Fundamental'), out='eppi_map.json'):
    next_id = 100_000

    # 1. Build Y-axis dimension (e.g. Region)
    y_vals = L(set(v for d in data for v in d['meta'].get(y_meta, []))).sorted()
    y_children = []
    for v in y_vals:
        a, next_id = mk_attr(v, next_id)
        y_children.append(a)
    y_dim, next_id = mk_attr(y_label, next_id, selectable=False, children=y_children)
    y_lookup = {a['AttributeName']: a['AttributeId'] for a in y_children}

    # 2. Build X-axis dimension (GCM objectives that appear with score >= min threshold)
    x_items = {}
    for d in data:
        for g in d['mappings'].get(theme, []):
            if g['relevance_score'] >= thresholds[0] and g['theme_id'] not in x_items:
                x_items[g['theme_id']] = g['theme_title']
    x_children = []
    for tid in sorted(x_items):
        a, next_id = mk_attr(f'Obj {tid}: {x_items[tid]}', next_id)
        x_children.append(a)
    x_dim, next_id = mk_attr(f'{theme.upper()} Objective', next_id, selectable=False, children=x_children)
    x_lookup = {tid: a['AttributeId'] for tid, a in zip(sorted(x_items), x_children)}

    seg_children = []
